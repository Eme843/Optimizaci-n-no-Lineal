# Optimización no lineal

Este cuaderno contiene la prueba inicial del proyecto y un ejemplo de uso de la clase `opt_unconstrained`.

```python
import numpy as np
from opt import opt_unconstrained


def f(x):
    return np.sum(x**2)


def grad(x):
    return 2 * x


x0 = np.array([3.0, -2.0])
solver = opt_unconstrained(f, grad, x0, nombre='Ejemplo')

print('f(x0) =', solver.f(x0))
print('grad(x0) =', solver.grad(x0))
print('evaluaciones f =', solver.nf)
print('evaluaciones grad =', solver.ng)
```

## Interpretación

La función objetivo es una forma cuadrática convexa. El gradiente es lineal, por lo que el ejemplo sirve para validar que la clase cuenta evaluaciones y devuelve valores coherentes con la teoría.
